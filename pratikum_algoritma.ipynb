```python
nilai = 82
kehadiran = 90
print("=== Kondisi Awal ===")
print("Nilai:", nilai, "| Kehadiran:", kehadiran)
syarat_nilai = nilai >= 75
syarat_kehadiran = kehadiran >= 80

print("\n--- Pemeriksaan Syarat ---")
print("Syarat Nilai (>= 75)     :", syarat_nilai)
print("Syarat Kehadiran (>= 80) :", syarat_kehadiran)

lulus_and = syarat_nilai and syarat_kehadiran
print("Status Lulus (menggunakan AND):", lulus_and)
print("\n" + "="*40 + "\n")
nilai = 70
syarat_nilai_baru = nilai >= 75
syarat_kehadiran_baru = kehadiran >= 80
lulus_and_baru = syarat_nilai_baru and syarat_kehadiran_baru

print("=== Setelah Nilai Diubah Menjadi 70 ===")
print("Nilai Baru:", nilai)
print("Syarat Nilai Baru (>= 75)     :", syarat_nilai_baru)
print("Syarat Kehadiran Baru (>= 80) :", syarat_kehadiran_baru)
print("Status Lulus (menggunakan AND):", lulus_and_baru)

# Menyiapkan variabel
nilai = 70
kehadiran = 90
# Evaluasi syarat
syarat_nilai = nilai >= 75
syarat_kehadiran = kehadiran >= 80
# Pengujian dengan operator OR
lulus_or = syarat_nilai or syarat_kehadiran
# Tampilkan hasil
print("Status Lulus (menggunakan OR):", lulus_or)
```

```python
#  Membuat variabel harga dan jumlah
harga = 12500
jumlah = 4
print ("harga per buku : rp ",harga)
print ("jumlah buku yang di beli :", jumlah )

#  Hitung total harga
total_harga = harga * jumlah
print ("total harga : rp", total_harga)
# hitung sisa pembagian
sisa_pembagian = total_harga % 2000
print ("sisa pembagian (% 2000):", sisa_pembagian)
# Hitung harga setelah diskon Rp5.000
harga_setelah_diskon = total_harga - 5000
print ("harga setelah diskon  : Rp", harga_setelah_diskon)
# Menampilkan hasil setiap perhitungan dengan label yang jelas
print("--- Hasil Perhitungan ---")
print("Total Harga             : Rp", total_harga)
print("Sisa Pembagian (% 2000) :", sisa_pembagian)
print("Harga Setelah Diskon    : Rp", harga_setelah_diskon)
```

```python
# Langkah 1: Simpan poin awal
poin = 100
print("Langkah 1 (Nilai awal) :", poin)

# Langkah 2: Tambahkan 25 poin
poin += 25
print("Langkah 2 (poin += 25) :", poin)

# Langkah 3: Kurangi 10 poin
poin -= 10
print("Langkah 3 (poin -= 10) :", poin)

# Langkah 4: Kalikan poin dengan 2
poin *= 2
print("Langkah 4 (poin *= 2)  :", poin)

# Langkah 5: Bagi poin dengan 2
poin /= 2
print("Langkah 5 (poin /= 2)  :", poin)
```

```python
nama = "Aress"
print("Nama        :", nama)
nilai_tugas = 80
nilai_ujian = 85
print("Nilai Tugas :", nilai_tugas)
print("Nilai Ujian :", nilai_ujian)
rata_rata = (nilai_tugas + nilai_ujian) / 2
print("Rata-rata   :", rata_rata)
status_lulus = rata_rata >= 75
print("Status Lulus:", status_lulus)
print("=== REKAP NILAI MAHASISWA ===")
print("Nama        :", nama)
print("Nilai Tugas :", nilai_tugas)
print("Nilai Ujian :", nilai_ujian)
print("Rata-rata   :", rata_rata)
print("Status Lulus:", status_lulus)
```
