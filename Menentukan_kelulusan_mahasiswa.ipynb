```python

# Program Menentukan Kelulusan Mahasiswa

nama = input("Masukkan nama mahasiswa: ")
nilai = float(input("Masukkan nilai akhir: "))

print("\n=== HASIL KELULUSAN ===")
print("Nama mahasiswa:", nama)
print("Nilai akhir:", nilai)

if nilai >= 60:
    print("Status: LULUS")
else:
    print("Status: TIDAK LULUS")
```
